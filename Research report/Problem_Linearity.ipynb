```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Style noir
# -----------------------------
plt.style.use("dark_background")

# -----------------------------
# Données simulées (illustratives)
# -----------------------------
np.random.seed(1)

# Axe X : stock level
x = np.linspace(0, 100, 120)

# Observations : forme sigmoïde + bruit
y_obs = 1 / (1 + np.exp(-(x - 35) / 6))
y_obs = y_obs + 0.05 * np.random.randn(len(x))

# Approche traditionnelle : relation linéaire
y_trad = 0.012 * x

# -----------------------------
# "Machine Learning" approach
# (courbe lisse non linéaire, type modèle flexible)
# -----------------------------
y_ml = 1 / (1 + np.exp(-(x - 35) / 7))

# -----------------------------
# Plot
# -----------------------------
fig, ax = plt.subplots(figsize=(7, 4))

# Observations
ax.scatter(
    x,
    y_obs,
    color="lightgray",
    s=18,
    alpha=0.8,
    label="Observations"
)

# Approche traditionnelle
ax.plot(
    x,
    y_trad,
    color="red",
    linewidth=2,
    label="Traditional approach"
)

# Approche Machine Learning
ax.plot(
    x,
    y_ml,
    color="lime",        # 👈 vert bien visible sur fond noir
    linewidth=2.5,
    label="Machine Learning approach"
)

# -----------------------------
# Mise en forme
# -----------------------------
ax.set_title("Stock management — Traditional vs Machine Learning", fontsize=12)
ax.set_xlabel("Stock level")
ax.set_ylabel("Fill rate")

ax.grid(alpha=0.2)
ax.legend(frameon=False)

plt.tight_layout()
plt.show()
```
